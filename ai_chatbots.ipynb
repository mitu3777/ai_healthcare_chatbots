```python
from transformers import pipeline, AutoModelForQuestionAnswering, AutoTokenizer
import gradio as gr
```

```python
model_name = "deepset/bert-base-cased-squad2"

# Load model and tokenizer
model = AutoModelForQuestionAnswering.from_pretrained(model_name, ignore_mismatched_sizes=True)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Initialize the pipeline
qa_pipeline = pipeline("question-answering", model=model, tokenizer=tokenizer)
```

```python
def healthcare_chatbot(question):
    context = """
    A fever is a temporary increase in body temperature, often due to an illness. 
    It is a sign that the body is fighting an infection. 
    Common symptoms include sweating, chills, headache, and muscle aches.

    A cold is a viral infection of your upper respiratory system. Symptoms include a runny nose, sore throat, coughing, and sneezing. The common cold is usually mild and resolves within a week or two.

    The flu (influenza) is also a viral infection but typically more severe than the common cold. Symptoms include fever, chills, muscle aches, cough, sore throat, and fatigue. It can last for up to two weeks and may require medical treatment.

    Headaches are common and can occur due to various reasons, including tension, dehydration, or underlying conditions like migraines. A severe headache, accompanied by other symptoms like nausea or sensitivity to light, may indicate a migraine.

    Diabetes is a chronic condition where the body is unable to regulate blood sugar levels properly. Symptoms include increased thirst, frequent urination, blurred vision, and fatigue. There are two main types: Type 1 and Type 2 diabetes. Type 2 is more common and often related to lifestyle factors.
    """
    # Use the pipeline to get the answer
    response = qa_pipeline(question=question, context=context)
    return response['answer']
```

```python
iface = gr.Interface(
    fn=healthcare_chatbot, 
    inputs="text", 
    outputs="text", 
    title="AI Healthcare Chatbot",
    description="Ask me a question"
)

# Launch the interface
iface.launch()
```
